```python
import requests
response = requests.get('https://github.com')
```

```python
# Option 1: Bracket notation
content_type_1 = response.headers['Content-Type']
```

```python
# Option 2: The .get() method (safer, prevents KeyError if missing)
content_type_2 = response.headers.get('Content-Type')
```

```python
print(content_type_1)
```

```python
print(content_type_2)
```
