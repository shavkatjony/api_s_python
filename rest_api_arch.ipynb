## getting into restAPI

### The REST architectural design 
- enables the client and server to be implemented independently of one another without being aware of each other.
- This means code on either side can be changed without worrying about how the change will affect the other.

```python
# Data from REST web services are exposed to the internet through a public URL,
# which can be accessed by sending an HTTP request.  
```

### http method  = description
- GET  	=  Retrieve data
- POST	=  Create data
- PUT	= Update existing data
- PATCH	 = Partially update existing data
- DELETE = Delete data
